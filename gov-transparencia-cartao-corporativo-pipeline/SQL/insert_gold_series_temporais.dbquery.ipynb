```sql
CREATE OR REPLACE TABLE gold_series_temporais
USING DELTA
AS
WITH base AS (
  SELECT
    XXHASH64(TRIM(UPPER(portador_nome))) AS id_portador,
    portador_nome AS nome,
    DATE_TRUNC('month', data_transacao) AS mes,
    YEAR(data_transacao) AS ano,
    valor_transacao,

    CASE MONTH(data_transacao)
      WHEN 1 THEN 'Janeiro'
      WHEN 2 THEN 'Fevereiro'
      WHEN 3 THEN 'Março'
      WHEN 4 THEN 'Abril'
      WHEN 5 THEN 'Maio'
      WHEN 6 THEN 'Junho'
      WHEN 7 THEN 'Julho'
      WHEN 8 THEN 'Agosto'
      WHEN 9 THEN 'Setembro'
      WHEN 10 THEN 'Outubro'
      WHEN 11 THEN 'Novembro'
      WHEN 12 THEN 'Dezembro'
    END AS mes_nome

  FROM silver.cartao_corporativo
)

SELECT
  id_portador,
  nome,
  ano,
  mes,
  mes_nome,

  CAST(SUM(valor_transacao) AS DECIMAL(18,2)) AS total_mes,
  COUNT(*) AS qtd_transacoes_mes,
  CAST(AVG(valor_transacao) AS DECIMAL(18,2)) AS media_mes

FROM base

GROUP BY id_portador, nome, ano, mes, mes_nome
ORDER BY total_mes DESC;
```
