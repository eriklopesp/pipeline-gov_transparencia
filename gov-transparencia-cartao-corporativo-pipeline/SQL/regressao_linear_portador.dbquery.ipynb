```sql

SELECT
  *,

  LAG(total_mes, 1) OVER (
    PARTITION BY id_hash_estabelecimento
    ORDER BY mes
  ) AS lag_1,

  LAG(total_mes, 3) OVER (
    PARTITION BY id_hash_estabelecimento
    ORDER BY mes
  ) AS lag_3,

  AVG(total_mes) OVER (
    PARTITION BY id_hash_estabelecimento
    ORDER BY mes
    ROWS BETWEEN 2 PRECEDING AND CURRENT ROW
  ) AS media_movel_3,

  CAST((total_mes - LAG(total_mes, 1) OVER (
    PARTITION BY id_hash_estabelecimento
    ORDER BY mes
  )) / NULLIF(
    LAG(total_mes, 1) OVER (
      PARTITION BY id_hash_estabelecimento
      ORDER BY mes
    ), 0
   * 100) AS DECIMAL (5,2)) AS crescimento_pct

FROM gov_transparencia.gold.gold_series_temporais_estabelecimentos;
```
